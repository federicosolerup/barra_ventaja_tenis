```python
config.pixel_width = 1920
config.pixel_height = 1080
config.frame_rate = 30

color_fondo_canvas = "#070B14"
color_fondo_lateral = "#0F172A"
color_borde_lateral = "#38BDF8"
color_sacador = "#10B981"
color_restador = "#3B82F6"
color_dorado = "#FBBF24"
color_texto_principal = "#F8FAFC"
color_texto_secundario = "#94A3B8"
fuente_texto = "Arial"

datos_tiros_animacion = [
    {"paso": 0, "desc": "0. Estado previo al saque", "hitter": "Pre-saque", "p_dp": 0.64, "p_fed": 0.36},
    {"paso": 1, "desc": "1. 1er Saque falta en red (DP)", "hitter": "Del Potro", "p_dp": 0.51, "p_fed": 0.49},
    {"paso": 2, "desc": "2. 2do Saque a la T (DP)", "hitter": "Del Potro", "p_dp": 0.55, "p_fed": 0.45},
    {"paso": 3, "desc": "3. Slice de revés al centro profundo (F)", "hitter": "Federer", "p_dp": 0.57, "p_fed": 0.43},
    {"paso": 4, "desc": "4. Derecha al centro corta (DP)", "hitter": "Del Potro", "p_dp": 0.50, "p_fed": 0.50},
    {"paso": 5, "desc": "5. Revés paralelo profundo (F)", "hitter": "Federer", "p_dp": 0.51, "p_fed": 0.49},
    {"paso": 6, "desc": "6. Derecha al centro corta (DP)", "hitter": "Del Potro", "p_dp": 0.47, "p_fed": 0.53},
    {"paso": 7, "desc": "7. Approach derecha cruzada media (F)", "hitter": "Federer", "p_dp": 0.29, "p_fed": 0.71},
    {"paso": 8, "desc": "8. Derecha paralela media (DP)", "hitter": "Del Potro", "p_dp": 0.16, "p_fed": 0.84},
    {"paso": 9, "desc": "9. Volea de revés cruzada corta ganadora (F)", "hitter": "Federer", "p_dp": 0.00, "p_fed": 1.00}
]

marcas_tiempo = ["01:14", "09:15", "10:11", "12:02", "13:07", "14:11", "15:21", "17:04", "17:22"]

def convertir_marca_a_segundos(marca_str):
    partes = marca_str.strip().split(":")
    return int(partes[0]) + int(partes[1]) / 30.0

tiempos_segundos = [0.0] + [convertir_marca_a_segundos(m) for m in marcas_tiempo]
duraciones_intervalos = [tiempos_segundos[i + 1] - tiempos_segundos[i] for i in range(len(tiempos_segundos) - 1)] + [4.267]

class BarraCanvasBroadcast(Scene):
    def construct(self):
        self.camera.background_color = color_fondo_canvas

        ancho_video = 12.24
        alto_video = 6.885
        pos_x_video = -0.83
        pos_y_video = 0.00

        ventana_video = Rectangle(
            width=ancho_video, height=alto_video,
            fill_color="#0B1120", fill_opacity=1.0,
            stroke_color="#38BDF8", stroke_width=2.5
        ).move_to(np.array([pos_x_video, pos_y_video, 0]))

        ancho_lateral = 1.55
        alto_lateral = alto_video
        pos_x_lateral = 6.20
        pos_y_lateral = pos_y_video

        caja_lateral = RoundedRectangle(
            corner_radius=0.18, width=ancho_lateral, height=alto_lateral,
            fill_color=color_fondo_lateral, fill_opacity=0.98,
            stroke_color=color_borde_lateral, stroke_width=2.0
        ).move_to(np.array([pos_x_lateral, pos_y_lateral, 0]))

        titulo_lateral = MarkupText("<b>EVALUACIÓN</b>", font=fuente_texto, font_size=11, color=color_dorado)
        titulo_lateral.move_to(np.array([pos_x_lateral, 3.18, 0]))

        ancho_columna = 1.05
        alto_barra = 3.70
        pos_y_barra = -0.08
        separacion = 0.18
        alto_tarjeta = 0.92

        y_techo_barra = pos_y_barra + alto_barra / 2
        y_piso_barra = pos_y_barra - alto_barra / 2

        marco_barra = RoundedRectangle(
            corner_radius=0.1, width=ancho_columna, height=alto_barra,
            stroke_color="#F8FAFC", stroke_width=1.8,
            fill_color="#0F172A", fill_opacity=1.0
        ).move_to(np.array([pos_x_lateral, pos_y_barra, 0]))

        linea_neutra = DashedLine(
            start=np.array([pos_x_lateral - ancho_columna / 2 - 0.05, pos_y_barra, 0]),
            end=np.array([pos_x_lateral + ancho_columna / 2 + 0.05, pos_y_barra, 0]),
            color="#FFFFFF", stroke_width=1.5, dash_length=0.06
        )

        y_centro_dp = y_techo_barra + separacion + alto_tarjeta / 2
        tarjeta_dp = RoundedRectangle(
            corner_radius=0.1, width=ancho_columna, height=alto_tarjeta,
            fill_color="#1E293B", fill_opacity=0.9,
            stroke_color=color_sacador, stroke_width=1.5
        ).move_to(np.array([pos_x_lateral, y_centro_dp, 0]))

        nombre_dp = MarkupText("<b>Del Potro</b>", font=fuente_texto, font_size=13.5, color=color_texto_principal)
        nombre_dp.move_to(tarjeta_dp.get_top() + DOWN * 0.24)

        y_centro_fed = y_piso_barra - separacion - alto_tarjeta / 2
        tarjeta_fed = RoundedRectangle(
            corner_radius=0.1, width=ancho_columna, height=alto_tarjeta,
            fill_color="#1E293B", fill_opacity=0.9,
            stroke_color=color_restador, stroke_width=1.5
        ).move_to(np.array([pos_x_lateral, y_centro_fed, 0]))

        nombre_fed = MarkupText("<b>Federer</b>", font=fuente_texto, font_size=13.5, color=color_texto_principal)
        nombre_fed.move_to(tarjeta_fed.get_bottom() + UP * 0.24)

        caja_superior = Rectangle(
            width=13.92, height=0.44,
            fill_color="#0B132B", fill_opacity=0.92,
            stroke_color="#1E293B", stroke_width=1.5
        ).move_to(np.array([0.0, 3.71, 0]))

        texto_encabezado = MarkupText(
            "<b>US OPEN 2009 — FINAL MASCULINA</b> | J.M. DEL POTRO vs. R. FEDERER",
            font=fuente_texto, font_size=13.5, color=color_dorado
        ).move_to(caja_superior.get_center())

        caja_inferior = Rectangle(
            width=13.92, height=0.46,
            fill_color="#0B132B", fill_opacity=0.95,
            stroke_color="#1E293B", stroke_width=1.5
        ).move_to(np.array([0.0, -3.71, 0]))

        rastreador_p = ValueTracker(datos_tiros_animacion[0]["p_dp"])

        def obtener_relleno_dp():
            p = rastreador_p.get_value()
            h = max(0.01, alto_barra * p)
            return RoundedRectangle(
                corner_radius=0.06, width=ancho_columna - 0.06,
                height=h, fill_color=color_sacador,
                fill_opacity=0.95, stroke_width=0
            ).move_to(np.array([pos_x_lateral, y_techo_barra - h / 2, 0]))

        def obtener_relleno_fed():
            p = rastreador_p.get_value()
            h = max(0.01, alto_barra * (1.0 - p))
            return RoundedRectangle(
                corner_radius=0.06, width=ancho_columna - 0.06,
                height=h, fill_color=color_restador,
                fill_opacity=0.95, stroke_width=0
            ).move_to(np.array([pos_x_lateral, y_piso_barra + h / 2, 0]))

        relleno_dinamico_dp = always_redraw(obtener_relleno_dp)
        relleno_dinamico_fed = always_redraw(obtener_relleno_fed)

        elementos_estaticos = Group(
            ventana_video, caja_lateral, titulo_lateral,
            tarjeta_dp, nombre_dp,
            marco_barra, linea_neutra,
            tarjeta_fed, nombre_fed,
            caja_superior, texto_encabezado,
            caja_inferior
        )

        self.add(elementos_estaticos)
        self.add(relleno_dinamico_dp, relleno_dinamico_fed)

        def construir_textos_paso(indice):
            d = datos_tiros_animacion[indice]
            pct_dp = int(round(d["p_dp"] * 100))
            pct_fed = int(round(d["p_fed"] * 100))

            txt_p_dp = MarkupText(f"<b>{pct_dp}%</b>", font=fuente_texto, font_size=23, color=color_sacador)
            txt_p_dp.move_to(tarjeta_dp.get_bottom() + UP * 0.28)

            txt_p_fed = MarkupText(f"<b>{pct_fed}%</b>", font=fuente_texto, font_size=23, color=color_restador)
            txt_p_fed.move_to(tarjeta_fed.get_top() + DOWN * 0.28)

            placa_paso = RoundedRectangle(
                corner_radius=0.08, width=1.50, height=0.30,
                fill_color="#1E293B", fill_opacity=0.95,
                stroke_color=color_dorado, stroke_width=1.0
            )
            txt_paso = MarkupText(f"<b>TIRO {d['paso']} / 9</b>", font=fuente_texto, font_size=10, color=color_dorado)
            txt_paso.move_to(placa_paso.get_center())
            grupo_placa = Group(placa_paso, txt_paso)

            txt_desc = MarkupText(f"<b>{d['desc']}</b>", font=fuente_texto, font_size=12.5, color=color_texto_principal)

            grupo_inferior = Group(grupo_placa, txt_desc).arrange(RIGHT, buff=0.30)
            grupo_inferior.move_to(caja_inferior.get_center())

            return Group(txt_p_dp, txt_p_fed, grupo_inferior)

        textos_actuales = construir_textos_paso(0)
        self.add(textos_actuales)
        self.wait(duraciones_intervalos[0])

        for i in range(1, len(datos_tiros_animacion)):
            nuevos_textos = construir_textos_paso(i)
            self.remove(textos_actuales)
            self.add(nuevos_textos)
            textos_actuales = nuevos_textos

            duracion_total = duraciones_intervalos[i]
            tiempo_animacion = min(0.40, duracion_total * 0.65)
            tiempo_reposo = max(0.0, duracion_total - tiempo_animacion)

            probabilidad_objetivo = datos_tiros_animacion[i]["p_dp"]
            self.play(rastreador_p.animate.set_value(probabilidad_objetivo), run_time=tiempo_animacion, rate_func=smooth)

            if tiempo_reposo > 0:
                self.wait(tiempo_reposo)

escena = BarraCanvasBroadcast()
escena.render()
```
