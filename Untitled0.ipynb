# Importar Dataset

```python
import os # Para usar funciones de sistema operativo como clonar
import pandas as pd

os.system("git clone https://github.com/federicosolerup/barra_ventaja_tenis.git") # Clonar el contenido del repo a este Colab
dataset_partidos = pd.read_csv('barra_ventaja_tenis/dataset/charting-m-matches.csv') # Acceder al csv que tiene los id de partidos
dataset_puntos = pd.read_csv('barra_ventaja_tenis/dataset/charting-m-points-modern.csv.zip',compression='zip') # Descomprimir el zip que tiene el csv de los puntos jugados

print('\n\n-PARTIDOS: Sólo sirve para separar train/test por partido según match_id y para eliminar partidos de zurdos según Pl_1_hand y Pl_2_hand.\n')
display(dataset_partidos.head(3))
print('\n\n-PUNTOS: Dataset principal que contiene la secuencia de golpes de cada punto jugado.\n')
display(dataset_puntos.head(3))
```

# Filtrar Dataset

```python
import re
from collections import Counter

for columna in ['Pl 1 hand', 'Pl 2 hand']:dataset_partidos[columna] = dataset_partidos[columna].astype(str).str.strip().str.upper() # Normalizar anotaciones de diestros "r" a "R"
dataset_partidos_diestros = dataset_partidos.loc[(dataset_partidos['Pl 1 hand'] == 'R') & (dataset_partidos['Pl 2 hand'] == 'R'),'match_id'] # Guardar los match_id de partidos entre jugadores diestros
dataset_puntos_filtrados = dataset_puntos[dataset_puntos['match_id'].isin(dataset_partidos_diestros)].copy() # Guardar puntos de partidos entre diestros
dataset_puntos_filtrados['secuencia_peloteo'] = dataset_puntos_filtrados['2nd'].fillna(dataset_puntos_filtrados['1st']) # Crear la columna secuencia_peloteo con la secuencia del saque (primero o segundo) que entró.

print(f'Puntos originales: {len(dataset_puntos):,}')
print(f'Puntos filtrados: {len(dataset_puntos_filtrados):,}')

descartes = Counter() # Contar cuántos puntos se descartan por cada motivo
validos = [] # Almacena los puntos aprobados

patron_saque = r'c*([0456])(\+?)'
patron_final = r'(?:[nwdxe][@#]|[@#*]|C)$'
patron_golpe = r'[fbrsvzopuylmhijktq][^fbrsvzopuylmhijktq]*'

for match_id, sacador, ganador, s in dataset_puntos_filtrados[['match_id', 'Svr', 'PtWinner', 'secuencia_peloteo']].itertuples(index=False, name=None):
    motivo = None
    if not isinstance(s, str) or not s.strip(): motivo = 'Puntos vacíos'
    else:
        s = s.strip()
        saque, final = re.match(patron_saque, s), re.search(patron_final, s)
        if any(c in s for c in 'SRPQVq'): motivo = 'Involucran al umpire'
        elif re.fullmatch(r'c*[0456]\+?[nwdxeg]#', s): motivo = 'Faltas de saque'
        elif re.fullmatch(r'c*[0456]\+?[nwdxeg]@?', s): motivo = 'Dobles faltas'
        elif not saque: motivo = 'Sin inicio válido'
        elif not final: motivo = 'Sin final válido'
        else:
            cuerpo = s[saque.end():final.start()]
            partes = re.findall(patron_golpe, cuerpo)
            if ''.join(partes) != cuerpo: motivo = 'Sin golpes'
            elif any(any(not (c.isdigit() or c in '+-=;^!') for c in p[1:]) for p in partes): motivo = 'Caracteres no válidos'
            elif any(sum(c in '0123' for c in p[1:]) > 1 or sum(c in '789' for c in p[1:]) > 1 for p in partes): motivo = 'Más de una dirección o profundidad'
            elif final.group() == 'C': motivo = 'Desafíos especiales'
            else:
                n = len(partes) + 1
                estimado = n % 2 == 1 if final.group() == '*' or (final.group() == '#' and n == 1) else n % 2 == 0
                if estimado != (sacador == ganador): motivo = 'Ganador inconsistente'
    if motivo: descartes[motivo] += 1
    else: validos.append((match_id, sacador, ganador, s))

dataset_puntos_validos = pd.DataFrame(validos, columns=['match_id', 'Svr', 'PtWinner', 'secuencia_peloteo'])

print(f'Puntos válidos: {len(dataset_puntos_validos):,}')
print(f'Puntos descartados: {sum(descartes.values()):,}\n')
print(*[f'- {k}: {v:,}' for k, v in descartes.most_common()], sep='\n')
```

# Parser

```python
def interpretar(p, saque=False):
    nums = re.findall(r'\d', p)
    direccion = next((n for n in nums if n in ('0456' if saque else '0123')), '-')
    profundidad = next((n for n in nums if n in '789'), '-') if not saque else '-'
    return ('saque' if saque else p[0], direccion if direccion != '0' else '-', profundidad, int('+' in p))

campos = ('golpe', 'direccion', 'profundidad', 'subida_a_red')
vacio = ('ninguno', '-', '-', 0)
ventanas = []

for match_id, sacador, ganador, s in dataset_puntos_validos.itertuples(index=False, name=None):
    saque = re.match(r'c*[0456]\+?', s)
    cuerpo = re.sub(r'(?:[nwdxe][@#]|[@#*]|C)$', '', s[saque.end():])
    tiros = [interpretar(saque.group(), True)] + [interpretar(p) for p in re.findall(r'[fbrsvzopuylmhijktq][^fbrsvzopuylmhijktq]*', cuerpo)]

    for i, tiro in enumerate(tiros):
        fila = dict(match_id=str(match_id), target=int(sacador == ganador), sacador_al_tiro=int(i % 2 == 0), indice_tiro=i)
        fila.update({pref + campo: valor for pref, j in [('anteprev_', i-2), ('prev_', i-1), ('', i)]
                     for campo, valor in zip(campos, tiros[j] if j >= 0 else vacio)})
        ventanas.append(fila)

dataset_tiros = pd.DataFrame(ventanas)
matriz_x = pd.get_dummies(dataset_tiros.drop(columns=['match_id', 'target'])) # Convertir categorías a columnas binarias.
vector_y = dataset_tiros['target'] # Variable objetivo: ganó o perdió el sacador
partidos_grupos = dataset_tiros['match_id'] # Agrupación por partido para evitar contaminación entre conjuntos.

print(matriz_x.shape)
display(matriz_x.head(3))

grupos = {
    'Anteanterior': [c for c in matriz_x if c.startswith('anteprev_')],
    'Anterior': [c for c in matriz_x if c.startswith('prev_')],
    'Actual': [c for c in matriz_x if not c.startswith(('anteprev_', 'prev_'))]
}

display(pd.DataFrame({k: pd.Series(v) for k, v in grupos.items()}).fillna(''))
```

# Train/Test

```python
from sklearn.model_selection import GroupShuffleSplit

# Separar los partidos para evitar contaminación entre entrenamiento y prueba.
division = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=42)
entrenamiento, prueba = next(division.split(matriz_x, vector_y, groups=partidos_grupos))

x_entrenamiento, x_prueba = matriz_x.iloc[entrenamiento], matriz_x.iloc[prueba]
y_entrenamiento, y_prueba = vector_y.iloc[entrenamiento], vector_y.iloc[prueba]

print(f'Entrenamiento: {len(x_entrenamiento):,} ventanas')
print(f'Prueba: {len(x_prueba):,} ventanas')
print(f'Partidos entrenamiento: {partidos_grupos.iloc[entrenamiento].nunique():,}')
print(f'Partidos prueba: {partidos_grupos.iloc[prueba].nunique():,}')
```

# Métricas

```python
import numpy as np
from sklearn.metrics import (accuracy_score, roc_auc_score, brier_score_loss,
                             precision_score, recall_score, f1_score)

def calcular_ece(y, prob, n=10):
    intervalos = np.linspace(0, 1, n + 1)
    ece = 0
    for i in range(n):
        mascara = (prob >= intervalos[i]) & (prob <= intervalos[i+1]) if i == 0 else (prob > intervalos[i]) & (prob <= intervalos[i+1])
        if mascara.any():
            ece += mascara.mean() * abs(prob[mascara].mean() - np.asarray(y)[mascara].mean())
    return ece

def evaluar_modelo(nombre, prob, pred):
    resultados = {'Modelo': nombre,
                  'Accuracy': f'{accuracy_score(y_prueba, pred):.2%}',
                  'AUC-ROC': f'{roc_auc_score(y_prueba, prob):.2%}',
                  'Brier Score': f'{brier_score_loss(y_prueba, prob):.4f}',
                  'ECE': f'{calcular_ece(y_prueba, prob):.2%}'}
    for clase, etiqueta in [(1, 'Sac.'), (0, 'Rest.')]:
        resultados.update({
            f'Prec. {etiqueta}': f'{precision_score(y_prueba, pred, pos_label=clase, zero_division=0):.2%}',
            f'Rec. {etiqueta}': f'{recall_score(y_prueba, pred, pos_label=clase, zero_division=0):.2%}',
            f'F1 {etiqueta}': f'{f1_score(y_prueba, pred, pos_label=clase, zero_division=0):.2%}'})
    return resultados

print('Funciones de evaluación definidas.')
```

# Entrenamiento

```python
import xgboost as xgb
from sklearn.metrics import confusion_matrix

columnas_t = [c for c in matriz_x if not c.startswith(('prev_', 'anteprev_'))]
columnas_t_prev = [c for c in matriz_x if not c.startswith('anteprev_')]
columnas_todas = matriz_x.columns.tolist()

parametros_xgb = dict(max_depth=8, learning_rate=0.05, n_estimators=200,
                      subsample=0.9, colsample_bytree=0.8, eval_metric='logloss',
                      random_state=42, tree_method='hist', device='cuda')

filas_ablacion = []

# Dummy Uniforme y Dummy Mayoría.
rng = np.random.default_rng(42)
prob = np.full(len(y_prueba), 0.5)
filas_ablacion.append(evaluar_modelo('Dummy Uniforme', prob, rng.integers(0, 2, len(y_prueba))))

p = y_entrenamiento.mean()
prob = np.full(len(y_prueba), p)
filas_ablacion.append(evaluar_modelo('Dummy Mayoría', prob, np.full(len(y_prueba), int(p >= 0.5))))

# Entrenar los tres modelos XGBoost.
for nombre, columnas in [('t', columnas_t), ('t, t-1', columnas_t_prev),
                         ('t, t-1, t-2', columnas_todas)]:
    modelo = xgb.XGBClassifier(**parametros_xgb)
    modelo.fit(x_entrenamiento[columnas], y_entrenamiento)
    prob = modelo.predict_proba(x_prueba[columnas])[:, 1]
    filas_ablacion.append(evaluar_modelo(f'XGBoost ({nombre})', prob, (prob >= 0.5).astype(int)))
    if nombre == 't, t-1, t-2':
        modelo_completo = modelo
        pred_completo = (prob >= 0.5).astype(int)

print('\n-MÉTRICAS:\n')
display(pd.DataFrame(filas_ablacion))

matriz = pd.DataFrame(
    confusion_matrix(y_prueba, pred_completo, labels=[0, 1]),
    index=['Real: Restador', 'Real: Sacador'],
    columns=['Predicho: Restador', 'Predicho: Sacador']
)

print('\n-MATRIZ DE CONFUSIÓN:\n')
display(matriz)

modelo_completo.save_model('modelo_xgboost_tenis.json')
print('\nModelo entrenado guardado en el archivo: modelo_xgboost_tenis.json')
```

# Caso de estudio

```python
partido = dataset_partidos[
    dataset_partidos['match_id'].str.contains('20090914.*US_Open.*Federer.*Del_Potro')
]['match_id'].iloc[0]

puntos_partido = dataset_puntos[dataset_puntos['match_id'] == partido]
punto = puntos_partido[puntos_partido['Gm#'] == 2].sort_values('Pt').iloc[0]

s = punto['2nd'] if pd.notna(punto['2nd']) else punto['1st']
saque = re.match(r'c*[0456]\+?', s)
cuerpo = re.sub(r'(?:[nwdxe][@#]|[@#*]|C)$', '', s[saque.end():])

codigos = [saque.group()] + re.findall(
    r'[fbrsvzopuylmhijktq][^fbrsvzopuylmhijktq]*', cuerpo
)

tiros = [interpretar(codigo, i == 0) for i, codigo in enumerate(codigos)]

filas = []
for i in range(len(tiros)):
    fila = {'sacador_al_tiro': int(i % 2 == 0), 'indice_tiro': i}
    fila.update({
        pref + campo: valor
        for pref, j in [('anteprev_', i-2), ('prev_', i-1), ('', i)]
        for campo, valor in zip(campos, tiros[j] if j >= 0 else vacio)
    })
    filas.append(fila)

x_punto = pd.get_dummies(pd.DataFrame(filas)).reindex(
    columns=columnas_todas, fill_value=0
)

prob_dp = modelo_completo.predict_proba(x_punto)[:, 1]
prob_f = 1 - prob_dp
```

```python
import matplotlib.pyplot as plt

x = np.arange(2, len(codigos) + 2)
etiquetas = [f'{n}\n{codigo}' for n, codigo in zip(x, codigos)]

fig, ax = plt.subplots(figsize=(13, 5))

for i, n in enumerate(x):
    ax.axvspan(n-.5, n+.5, color='lightcoral' if i % 2 == 0 else 'lightskyblue', alpha=.10)

ax.plot(x, prob_dp, 'o-', lw=2, color='tab:blue', label='Del Potro (sacador)')
ax.plot(x, prob_f, 's-', lw=2, color='tab:orange', label='Federer (restador)')
ax.axhline(.5, color='gray', ls='--', alpha=.6)

for i, (dp, f) in enumerate(zip(prob_dp, prob_f)):
    for valor, color in [(dp, 'tab:blue'), (f, 'tab:orange')]:
        ax.annotate(f'{valor:.1%}', (x[i], valor),
                    xytext=(0, 12 if valor >= .5 else -22),
                    textcoords='offset points', ha='center', color=color,
                    bbox=dict(boxstyle='round,pad=.2', fc='white', ec=color, alpha=.5))

ax.set(xticks=x, xticklabels=etiquetas, ylim=(-.05, 1.08),
       xlabel='Número de tiro y código registrado',
       ylabel='Probabilidad de ganar el punto',
       title='US Open 2009 — Del Potro vs. Federer')
ax.grid(alpha=.2)
ax.legend()

plt.tight_layout()
plt.savefig('eval_bar_simulation.png', dpi=300)
plt.show()
```

# Tiro 4

```python
import json
from scipy.special import expit, logit

booster = modelo_completo.get_booster()
arboles = [json.loads(a) for a in booster.get_dump(dump_format='json')]
x = x_punto.iloc[2]  # Cuarto tiro del gráfico (posición 4).

# Recorrer el primer árbol.
nodo = arboles[0]
recorrido = []

while 'leaf' not in nodo:
    variable = nodo['split']
    valor = float(x[variable])
    siguiente = nodo['yes'] if valor < nodo.get('split_condition', 1) else nodo['no']
    recorrido.append((variable, valor, siguiente))
    nodo = next(h for h in nodo['children'] if h['nodeid'] == siguiente)

display(pd.DataFrame(recorrido, columns=['Variable', 'Valor', 'Siguiente nodo']))

# Sumar las contribuciones de los 200 árboles.
pesos = []
for arbol in arboles:
    nodo = arbol
    while 'leaf' not in nodo:
        variable = nodo['split']
        siguiente = nodo['yes'] if float(x[variable]) < nodo.get('split_condition', 1) else nodo['no']
        nodo = next(h for h in nodo['children'] if h['nodeid'] == siguiente)
    pesos.append(float(nodo['leaf']))

base = json.loads(booster.save_config())['learner']['learner_model_param']['base_score']
sesgo = logit(float(base.strip('[]')))
total = sesgo + sum(pesos)

display(pd.DataFrame({
    'Componente': ['Sesgo base', 'Árbol 1', 'Suma de 200 árboles', 'Logit final', 'Probabilidad final'],
    'Valor': [sesgo, pesos[0], sum(pesos), total, expit(total)]
}))

print('Probabilidad predict_proba:', modelo_completo.predict_proba(x_punto.iloc[[2]])[0, 1])
```
